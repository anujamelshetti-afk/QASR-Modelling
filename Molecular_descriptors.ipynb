```python
from rdkit import Chem
from rdkit.Chem import Descriptors
import pandas as pd
print("RDKit is working!")
```

```python
ic50=pd.read_csv('bace1_clean.csv')
```

```python
ic50["Mol"] = ic50["Smiles"].apply(Chem.MolFromSmiles)

ic50.head()
```

```python
print(ic50["Mol"].isnull().sum())
```

```python
ic50["MolWt"] = ic50["Mol"].apply(Descriptors.MolWt)
ic50["LogP"] = ic50["Mol"].apply(Descriptors.MolLogP)
ic50["TPSA"] = ic50["Mol"].apply(Descriptors.TPSA)
ic50["HBD"] = ic50["Mol"].apply(Descriptors.NumHDonors)
ic50["HBA"] = ic50["Mol"].apply(Descriptors.NumHAcceptors)
ic50["RotBonds"] = ic50["Mol"].apply(Descriptors.NumRotatableBonds)

print("Descriptors calculated!")
```

```python
ic50[
    [
        "Smiles",
        "MolWt",
        "LogP",
        "TPSA",
        "HBD",
        "HBA",
        "RotBonds",
        "pIC50"
    ]
].head(25)
```

```python
ic50.to_csv("bace1_descriptors.csv", index=False)
```
