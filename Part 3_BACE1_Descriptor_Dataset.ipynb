```python
# check for invalid smiles
from rdkit import Chem

invalid = []

for i, smi in enumerate(df3['canonical_smiles']):
    if Chem.MolFromSmiles(smi) is None:
        invalid.append(i)

print("Invalid molecules:", len(invalid))
print(invalid[:20])
```

```python
batch_size = 100

for i in range(0, len(df3_selection), batch_size):
    batch = df3_selection.iloc[i:i+batch_size]
    batch.to_csv(
        f"batch_{i}.smi",
        sep="\t",
        index=False,
        header=False
    )
```

```python
from padelpy import padeldescriptor

failed_batches = []

for i in range(0, len(df3_selection), 100):
    print(f"Running batch {i}")

    try:
        padeldescriptor(
            mol_dir=f"batch_{i}.smi",
            d_file=f"batch_{i}.csv",
            fingerprints=True,
            removesalt=True,
            threads=2
        )
    except Exception as e:
        print(f"Batch {i} failed")
        failed_batches.append(i)

print("Failed batches:", failed_batches)
```

```python
from padelpy import padeldescriptor

failed = []

for i in range(2600, 2700):

    one = df3_selection.iloc[[i]]

    one.to_csv(
        "one.smi",
        sep="\t",
        index=False,
        header=False
    )

    try:
        padeldescriptor(
            mol_dir="one.smi",
            d_file="one.csv",
            fingerprints=True,
            removesalt=True,
            threads=1
        )

    except Exception as e:
        print(f"Failed row: {i}")
        print(df3_selection.iloc[i]["molecule_chembl_id"])
        print(e)
        failed.append(i)

print("Failed rows:", failed)
```
