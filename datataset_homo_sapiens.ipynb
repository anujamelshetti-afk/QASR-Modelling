```python
# keep only IC50 measurements
ic50 = df[df['Standard Type'] == "IC50"].copy()

print("Number of IC50 records:", len(ic50))
```

```python
ic50 = ic50[ic50["Standard Relation"] == "="]
```

```python
ic50 = ic50[
    [
        "Molecule ChEMBL ID",
        "Smiles",
        "Standard Value",
        "Standard Units",
        "pChEMBL Value"
    ]
]

ic50.head()
```

```python
ic50 = ic50.dropna()

print(ic50.shape)
```

```python
ic50 = ic50.drop_duplicates(subset="Smiles")

print(ic50.shape)
```

```python
ic50.isnull().sum()
```

```python
ic50 = ic50.dropna()
print(ic50.shape)
```

```python
ic50[["Standard Value", "pChEMBL Value"]].head(10)
```

```python
ic50 = ic50.rename(columns={"pChEMBL Value": "pIC50"})
```
